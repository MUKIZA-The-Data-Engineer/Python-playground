```python
import pandas as pd

# 1. Identify all text/categorical columns automatically
text_cols = df.select_dtypes(include=['object', 'string']).columns

# 2. Apply vectorized string transformations across all identified columns
for col in text_cols:
    df[col] = (
        df[col]
        .astype(str)                   # Ensures data type is string
        .str.strip()                    # Removes leading/trailing spaces
        .str.replace(r'\s+', ' ', regex=True) # Replaces multiple spaces with a single space
        .str.capitalize()               # Capitalizes 1st letter, makes rest lowercase
    )

# 3. Clean up any literal 'nan' strings created during astype(str) conversion
#import numpy as np
df[text_cols] = df[text_cols].replace({'Nan': np.nan, 'nan': np.nan, 'None': np.nan})
```

```python
for col in ['sales', 'quantity', 'discount', 'profit']:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(r'[$]', '', regex=True)
        .str.strip()
    )

# Convert to numeric
df[['sales', 'quantity', 'discount', 'profit']] = df[['sales', 'quantity', 'discount', 'profit']].apply(pd.to_numeric, errors='coerce')
```

```python
import numpy as np
import pandas as pd


def standardize_text_columns(
    df: pd.DataFrame, casing: str = "capitalize"
) -> pd.DataFrame:
  """Standardizes all text/object columns in a DataFrame:

  1. Trims outer whitespace
  2. Replaces inner multi-spacing with a single space
  3. Standardizes casing ('capitalize', 'title', 'lower', or 'upper')
  4. Preserves missing values (NaN)
  """
  df = df.copy()  # Prevents modifying the original input unexpectedly
  text_cols = df.select_dtypes(include=["object", "string"]).columns

  for col in text_cols:
    cleaned_series = (
        df[col].astype(str).str.strip().str.replace(r"\s+", " ", regex=True)
    )

    if casing == "capitalize":
      cleaned_series = cleaned_series.str.capitalize()
    elif casing == "title":
      cleaned_series = cleaned_series.str.title()
    elif casing == "lower":
      cleaned_series = cleaned_series.str.lower()
    elif casing == "upper":
      cleaned_series = cleaned_series.str.upper()

    # Re-convert string 'nan' entries back to true Nulls
    df[col] = cleaned_series.replace(
        {"Nan": np.nan, "nan": np.nan, "None": np.nan}
    )

  return df
```
